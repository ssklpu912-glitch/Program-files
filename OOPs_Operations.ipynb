```python
class fraction():           

    def __init__(self, n, d):  ## parameterised constructor  ##  magic methods
        self.num=n
        self.den=d

    def __str__(self):           ##  how it should excecute   ##  magic methods
        return "{}/{}".format(self.num, self.den)

    def __add__(self, other): 

        temp_num=(self.num*other.den + self.den*other.num)
        temp_den=self.den*other.den
        return "{}/{}".format(temp_num, temp_den)

    def __sub__(self, other): 

        temp_num2=(self.num*other.den - self.den*other.num)
        temp_den2=self.den*other.den
        return "{}/{}".format(temp_num2, temp_den2)

    def __mul__(self, other): 

        temp_num3=self.num*other.num
        temp_den3=self.den*other.den
        return "{}/{}".format(temp_num3, temp_den3)

    def __truediv__(self, other): 

        temp_num4=self.num*other.den
        temp_den4=self.den*other.num
        return "{}/{}".format(temp_num4, temp_den4)

    def con_to_deci(self):                      ##  non magic methods
        return self.num/self.den
        

fr1=fraction(3,4)
fr2=fraction(7,4)
```

```python
print(fr1+fr2)
print(fr1-fr2)
print(fr1*fr2)

print(fr1/fr2)
fr1.con_to_deci()
```

```python
class point():

    def __init__(self, x, y):
        self.xcod=x
        self.ycod=y

    def __str__(self):
        return "{},{}".format(self.xcod, self.ycod)

    def euclid_dis(self, other):
        return ((self.xcod -other.xcod)**2-(self.ycod -other.ycod)**2)**0.5

    def pythag_hypo(self):
        return ((self.xcod)**2+(self.ycod)**2)**0.5

class line():
    
    def __init__(self, A, B, C):   ###  x+y+c=0
        self.A=A
        self.B=B
        self.C=C

    def __str__(self):
        return "{}+{}+{}=0".format(self.A, self.A, self.C)

    def point_on_line(line, point):
        if line.A*point.xcod + line.B*point.ycod + line.C==0:
            return 'lies on the line'
        else:
            return 'does not lies on the line'

        
```
