```python
# Q9: Extract the sensor ID and current value.
import re

log = "Time: 12:05:33 | Sensor: S3 | Value: 3.45 mA"

# Extract sensor ID and value using regular expressions
sensor_match = re.search(r'Sensor:\s*(S\d+)', log)
value_match = re.search(r'Value:\s*([\d.]+)', log)

Sensor_ID = sensor_match.group(1) if sensor_match else None
current_value = float(value_match.group(1)) if value_match else None

print("The sensor ID is:", Sensor_ID)
print("The current value is:", current_value)
```

```python
# A list of filenames:
files = ["sample_A1.txt", "sample_A2.txt", "control_B1.txt", "control_B2.txt"]

# Q11: Count how many are sample vs control files.
sample_count = 0
control_count = 0

for file in files:
    if "sample" in file:
        sample_count+= 1
    elif "control" in file:
        control_count+= 1

print("the number of sample files are:", sample_count)
print("the number of control files are:", control_count)       

```

```python
#Extract the well IDs (e.g., A1, A2, B1, B2).
files = ["sample_A1.txt", "sample_A2.txt", "control_B1.txt", "control_B2.txt"]

well_IDs = []

for filename in files:
    match = re.search(r'_(\w\d)\.txt$', filename)  # or r'_([A-Z]\d+)' for more general
    if match:
        well_IDs.append(match.group(1))

print("The well IDs are:", well_IDs)
```

```python
#Q14: Split the string by parameter values (e.g., "Temp: 37C", "pH=7.4", "Reading: 0.89").
note = "Temp: 37C...pH=7.4////Reading: 0.89  "
clean_1 = re.sub(r'\.{2,}|/{2,}', '', note)
clean_1 = re.sub(r'\s+', ' ', clean_1).strip()
Temp_match = re.search(r'Temp:\s*(\d+C)', clean_1)
pH_match = re.search(r'pH=\s*\d+\.\d+', clean_1)
Reading_match = re.search(r'Reading:\s*\d+\.\d+', clean_1)
temp = f"Temp = {Temp_match.group(0)}" if Temp_match else "Temp: = N/A"
pH = f"pH = {pH_match.group(0)}" if pH_match else "pH = N/A"
reading = f"Reading = {Reading_match.group(0)}" if Reading_match else "Reading: = N/A"
print(temp)
print(pH)
print(reading)
```

```python
#Given a list of wavelengths and absorbance values, write a conditional to:
#Identify if a peak occurs at 530 ± 5 nm #Raise a flag if more than one peak is detected in that range
spectral_value = [
    {'w': 525, 'abs': 1.2},
    {'w': 525, 'abs': 1.0},
    {'w': 530, 'abs': 0.2},
    {'w': 535, 'abs': 1.5},
    {'w': 535, 'abs': 1.1}
]

# Filter peaks within 530 ± 5 nm range
peaks_in_range = [d for d in spectral_value if 525 <= d['w'] <= 535]

if len(peaks_in_range) == 1:
    print("Peak occurs at:", peaks_in_range)
elif len(peaks_in_range) > 1:
    print("Multiple peaks occur at:", peaks_in_range)
else:
    print("No peaks in the range")
```
