### Date_range function

```python
##  give a range of dates from 'start' to 'end'

pd.date_range(start='2026-07-14', end='2026-07-31', freq='D')
```

```python
pd.date_range(start='2026-07-14', end='2026-07-31', freq='2D')  ###  '2D' gives the alternate dates
```

```python
###  only buisness days-------> no weekend days

pd.date_range(start='2026-07-14', end='2026-07-31', freq='B')
```

```python
## only the one day of a week

pd.date_range(start='2026-07-14', end='2026-07-31', freq='W')
```

```python
pd.date_range(start='2026-07-14', end='2026-07-31', freq='W-THU')   ###  only thursdays
```

```python
##  hourly date stamps

pd.date_range(start='2026-07-14', end='2026-07-31', freq='h')
```

```python
###  for every 6 hours

pd.date_range(start='2026-07-14', end='2026-07-31', freq='6h')
```

```python
### only month end date

pd.date_range(start='2026-07-14', end='2026-07-31', freq='ME')
```

```python
###  only month start date

pd.date_range(start='2026-07-14', end='2026-07-31', freq='ME')
```

```python
###  only year end date

pd.date_range(start='2026-07-14', end='2032-07-31', freq='YE')
```

```python
###  using periods(number of results)

pd.date_range(start='2026-07-14', periods=25, freq='D')
```

```python
###  using periods(number of results)----> hourly

pd.date_range(start='2026-07-14', periods=25, freq='h')
```

```python
###  using periods(number of results)-----> every 6hours

pd.date_range(start='2026-07-14', periods=25, freq='6h')
```
