# Relational plots

```python
###  seaborn functions 

### scatterplot func-----> axes level function

sns.scatterplot(data=P_record, x='age', y='height_cm')   ### no need to write te labels separately like matplotlib
```

```python
###  we can add gender by using 'hue' attribute

sns.scatterplot(data=P_record, x='age', y='height_cm', hue='gender')  
```

```python
### for the fig level function----> we need to use 'relplot' function instead of 'scatterplot'
###  here we mostly get squarish shape graphs

sns.relplot(data=P_record, x='age', y='height_cm', kind='scatter', 
            hue='gender')
```

```python
sns.relplot(data=P_record, x='age', y='height_cm', kind='scatter', 
            hue='gender', size='age')
```

```python
###  facet plot ----> "only works with figure level graph"
```

```python
###  facet plot ----> making graph for each category ----->  use 'col' instead of hue

sns.relplot(data=P_record, x='age', y='height_cm', kind='scatter', col='gender', size='age')
```

```python
###  ploting for cols

sns.relplot(data=P_record, x='age', y='height_cm', kind='scatter', col='blood_type', size='age')
```

```python
###  ploting for rows

sns.relplot(data=P_record, x='age', y='height_cm', kind='scatter', row='blood_type', size='age')
```

```python
###  ploting for rows and cols both

sns.relplot(data=P_record, x='age', y='height_cm', kind='scatter', row='blood_type', col='smoking_status', size='age')
```

```python
###  ploting for rows and cols both ------->  line plot

sns.relplot(data=P_record, x='age', y='height_cm', kind='line', row='blood_type', col='smoking_status', size='age')
```

```python
###   col wrap----------> if there are more cols, then we can wrap number of cols per row

sns.relplot(data=P_record, x='age', y='height_cm', kind='scatter', row='blood_type', col='age', size='age', col_wrap=3)
```

```python
####  how to add stlyes  -----> use 'style'

sns.scatterplot(data=P_record, x='age', y='height_cm', hue='gender', style='smoking_status') 
```

```python
sns.scatterplot(data=P_record, x='age', y='height_cm', hue='gender', 
                style='blood_type') 
```

```python
####  how to add size parameter  -----> use 'size'

sns.scatterplot(data=P_record, x='weight_kg', y='height_cm', hue='gender', 
                style='smoking_status', size='age') 
```

## Pairplotting  the relations

```python
### all the daigonal plots are between the same x and y===== they are all histograms

sns.pairplot(P_record)
```

```python
sns.pairplot(P_record, hue='gender')
```

```python
###  pairgrid-----> which is behind plotting the pairplot

g1=sns.PairGrid(data=P_record)
g1.map(sns.scatterplot)
```

```python
###  plotting the daigonal items separately

g1=sns.PairGrid(data=P_record)
g1.map_diag(sns.scatterplot)
g1.map_offdiag(sns.histplot)
```

###  jointplot and jointgrid

```python
sns.jointplot(data=P_record, x='age', y='height_cm')
```

```python
sns.jointplot(data=P_record, x='age', y='height_cm', hue='gender')
```

```python
sns.jointplot(data=P_record, x='age', y='height_cm', hue='gender', kind='hist')
```

```python
###  Jointgrid func

g2=sns.JointGrid(data=P_record, x='height_cm', y='age')
g2.plot(sns.scatterplot, sns.histplot)
```

```python
##  can do any----> boxplot, kdeplot etc

g2=sns.JointGrid(data=P_record, x='height_cm', y='age')
g2.plot(sns.kdeplot, sns.boxplot)
```
