```python
### CREATING A CLASS

class fraction():

    def __init__(self, n, d):
        self.num=n                #### INSTANCE VARIABLES
        self.den=d

    def __str__(self):
        return "{}/{}.format(self.num, self.den)"

    def __add__(self, other):

        temp_num=(self.num*other.den + self.den*other.num)
        temp_den=self.den*other.den
        return "{}/{}.format(temp_num, temp_den)"

    def __sub__(self, other):

        temp_num=(self.num*other.den - self.den*other.num)
        temp_den=self.den*other.den
        return "{}/{}.format(temp_num, temp_den)"

    def __mul__(self, other):

        temp_num=(self.num*other.num)
        temp_den=self.den*other.den
        return "{}/{}.format(temp_num, temp_den)"

    def __truediv__(self, other):

        temp_num=(self.num*other.den)
        temp_den=self.den*other.num
        return "{}/{}.format(temp_num, temp_den)"
        
```

```python
class Atm():
    __counter = 1      #### STATIC VARIABLE
    
    def __init__(self):

        self._pin=1265                   ### INSTANCE VARIABLES ARE INSIDE THE CONSTRUCTOR    "__init__" 
        self._balance=52333           ### ____ THESE ARE ACCESSED BY self(object name)
        self.sno=0
        self.sno=Atm.__counter    #### FOR STATIC VARIABLES they are accessed by class name "Atm" 
        Atm.__counter=Atm.__counter+1

    @staticmethod        
    def get_counter():          ####  the function for static variable does not pass self
        return Atm.__counter

    @staticmethod 
    def set_counter(new):
        if type(new)==int:
            Atm.__counter=new
        else:
            print("not allowed")
        

    def menu(self):
        user_input=input("""Hello, how you would like to proceed?
        1. Enter 1 to create pin
        2. Enter 2 to check balance
        3. Enter 3 to withdraw cash
        4. Enter 4 to deposit cash
        5. Enter 5 to exit""")
        if user_input=="1":
            self.create_pin()
        elif user_input=="2":
            self.check_balance()
        elif user_input=="3":
            self.withdraw_cash()  
        elif user_input=="4":
            self.deposit_cash()
        else:
            print("proceed to exit")
            
    def create_pin(self):
        self._pin=int(input("enter a 4 digit pin"))
        print("pin set")

    def withdraw_cash(self):
        temp=int(input("enter your 4 digit pin"))
        if temp==self._pin:
            amount=int(input("enter an amount"))
            if amount<self._balance:
                self._balance=self._balance-amount
                print("your transaction is in progress")
            else:
                print("insufficient funds") 
        else:
            print("invalid pin")

c1 = Atm()
c2 = Atm()
c3 = Atm()

print(Atm.get_counter()) 
```

```python
##########  SUPER keyword   EXAMPLE

class Phone:

    def __init__(self, price, brand, camera):
        print("price brand")
        self.price=price 
        self.brand=brand 
        self.camera=camera

    def buy(self):                          
        print("buying a phone")

class smartphone(Phone):
    def buy(self):
        print("buying a smartphone")
        super().buy()                    ##### SUPER can call the buy() func of the parent class here

##@@@ --- this can only access parent clas "methods & constructor" ----------"not the attributes"

s=smartphone(52000, "Apple", "13mP")
s.buy()
```

```python
class Book:
    def __init__(self, book_id, title, subject):
        self.book_id = book_id
        self.title = title
        self.subject = subject
        self.is_issued = False
        self.issued_to = None      # will hold a member_id when issued
        self.issue_date = None
        self.return_date = None

    def __str__(self):
        status = "Issued" if self.is_issued else "Available"
        return f"[{self.book_id}] {self.title} ({self.subject}) - {status}"


class Member:
    def __init__(self, member_id, name, email):
        self.member_id = member_id
        self.name = name
        self.email = email
        self.borrowed_books = []      # list of book_ids currently held

    def __str__(self):
        return f"Member[{self.member_id}] {self.name}"


class Library:
    def __init__(self):
        self.books = {}      # book_id -> Book object
        self.members = {}    # member_id -> Member object
        self.menu()

    # ---------- setup helpers ----------
    def add_book(self, book_id, title, subject):
        self.books[book_id] = Book(book_id, title, subject)
        print(f"Book '{title}' added with ID {book_id}")

    def register_member(self, member_id, name, email):
        self.members[member_id] = Member(member_id, name, email)
        print(f"Member '{name}' registered with ID {member_id}")

    # ---------- main menu ----------
    def menu(self):
        while True:
            user_input = input("""
Welcome to the Library
1. Search a book
2. Issue a book
3. Return a book
4. Exit
Enter choice: """)
            if user_input == "1":
                self.search_book()
            elif user_input == "2":
                self.issue_book()
            elif user_input == "3":
                self.return_book()
            elif user_input == "4":
                print("Goodbye!")
                break
            else:
                print("Invalid option, try again")

    # ---------- actions ----------
    def search_book(self):
        keyword = input("Enter subject or title to search: ").lower()
        results = [b for b in self.books.values()
                   if keyword in b.title.lower() or keyword in b.subject.lower()]
        if results:
            for b in results:
                print(b)
        else:
            print("No matching books found")

    def issue_book(self):
        member_id = input("Enter your member ID: ")
        book_id = input("Enter the book ID to issue: ")

        member = self.members.get(member_id)
        book = self.books.get(book_id)

        if member is None:
            print("Invalid member ID")
            return
        if book is None:
            print("Book not found")
            return
        if book.is_issued:
            print("Book not available, added to waitlist (not yet implemented)")
            return

        book.is_issued = True
        book.issued_to = member_id
        member.borrowed_books.append(book_id)
        print(f"Book '{book.title}' issued to {member.name}")

    def return_book(self):
        member_id = input("Enter your member ID: ")
        book_id = input("Enter the book ID to return: ")

        member = self.members.get(member_id)
        book = self.books.get(book_id)

        if member is None or book is None:
            print("Invalid member or book ID")
            return
        if book.issued_to != member_id:
            print("This book was not issued to this member")
            return

        book.is_issued = False
        book.issued_to = None
        member.borrowed_books.remove(book_id)
        print(f"Book '{book.title}' returned by {member.name}")


# --- Usage ---
lib = Library()
```
