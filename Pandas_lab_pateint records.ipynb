```python
### Using each patient's height and weight, calculate their Body Mass Index (BMI). Then group patients 
##into standard categories (Underweight, Normal, Overweight, Obese) — how many patients fall into each category?

P_record['BMI']=round(P_record['weight_kg']/(P_record['height_cm']/100)**2)
P_record.head(1)
```

```python
print(P_record['weight_kg'].isna().sum())
```

```python
def BMI_category(BMI):
    
    if pd.isna('BMI'):
        return 'unknown'
    elif BMI < 18.5:
        return 'underweight'
    elif BMI < 24.9:
        return 'Healthy'
    elif BMI < 29.9:
        return 'overweight'
    else:
        return 'Obese'

P_record['category']=P_record['BMI'].apply(BMI_category)
P_record[['BMI', 'category']].head()
```

```python
def age_category(age):
    if age < 10:
        return 'toddler'
    elif age < 35:
        return 'younger_p'
    else:
        return 'older'

P_record['affected_population']= P_record['age'].apply(age_category)
P_record.head(5)
```
